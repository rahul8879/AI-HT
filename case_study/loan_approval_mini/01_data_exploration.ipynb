```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Q1) how many data point we have
df = pd.read_csv('loan_prediction.csv')


def basic_data_details(df):
    data_size = df.shape[0]
    data_info = df.info()
    null_cnt = df.isnull().sum()
    return data_size,data_info,null_cnt

basic_data_details(df)


```
