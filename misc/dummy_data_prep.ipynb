```python
import numpy as np

# Example data (replace zeros in A with something non-zero for demo)
A = np.array([[1, 2],
              [3, 4],
              [5, 6]])

B = np.array([[0.1, 0.5],
              [0.8, 0.2],
              [0.3, 0.9]])

print("A =\n", A)
print("B =\n", B)

# Compute C = A @ B.T
C = A @ B.T

print("\n--- Step-by-step computation of C = A @ B.T ---")
for i in range(A.shape[0]):      # for each row of A
    for j in range(B.shape[0]):  # for each row of B (or column of B.T)
        row_A = A[i, :]
        row_B = B[j, :]
        products = [f"{row_A[k]}*{row_B[k]}" for k in range(len(row_A))]
        result = row_A @ row_B
        print(f"C[{i},{j}] = ({' + '.join(products)}) = {result:.2f}")

print("\nFinal C =\n", C)
```

```python
#create a list of 4 reviews
reviews = [
    'The the product is great!',
    'I did not like the product.',
    'Excellent quality, will buy again.',
    'Not worth worth the price.'
]

# split it  and then apply sets to get the unqiue value
unique_words = set()
for review in reviews:
    unique_words.update(review.lower().split())
```

```python
len(unique_words)
```

```python
# lets create vector for each reviews
# create a vector size of unique words
import numpy as np

# Create a document-term matrix
doc_term_matrix = np.zeros((len(reviews), len(unique_words)))

# Fill the document-term matrix
for i, review in enumerate(reviews):
    for word in review.lower().split():
        if word in unique_words:
            doc_term_matrix[i, list(unique_words).index(word)] += 1
```

```python
import matplotlib.pyplot as plt

# Data: Advertising Spend (X) and Sales (Y)
advertising_spend = [100, 200, 300,400]
sales = [10, 20, 30,45]

# Predict line (simple linear relationship: Sales = 0.1 * Spend)
x_line = [0, 400]  # Range of ad spend
y_line = [0.1 * x for x in x_line]

# Plot data points
plt.scatter(advertising_spend, sales, color='blue', label='Actual Sales')

# Plot trend line
plt.plot(x_line, y_line, color='red', label='Predicted Line: Sales = 0.1*Spend')

# Labels and Title
plt.xlabel('Advertising Spend ($)')
plt.ylabel('Sales (in $1000s)')
plt.title('Retail Example: Advertising Spend vs. Sales')
plt.legend()
plt.grid(True)

# Show plot
plt.show()
```
